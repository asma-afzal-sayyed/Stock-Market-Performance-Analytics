```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import glob
import warnings

warnings.filterwarnings("ignore")

print("All libraries imported successfully!")
print("Pandas Version:", pd.__version__)
print("NumPy Version:", np.__version__)
```

```python

import pandas as pd
import os
import glob

# Dataset folder location
dataset_path = r"D:\Asma Portfolio\Stock Market Performance Analytics\dataset"

# Find all CSV files
csv_files = glob.glob(os.path.join(dataset_path, "*.csv"))

# Load all company datasets
stock_data = {}

for file in csv_files:
    company_name = os.path.basename(file).replace(".csv", "")
    stock_data[company_name] = pd.read_csv(file)

    print(f"{company_name} loaded successfully!")

print("\nTotal companies loaded:", len(stock_data))
```

```python
# Display first 5 rows of each company dataset

for company, df in stock_data.items():
    print("\nCompany:", company)
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    display(df.head())
```

```python
# Check date format and missing values

for company, df in stock_data.items():
    print("\nCompany:", company)
    print("Missing Values:")
    print(df.isnull().sum())
    print("\nData Types:")
    print(df.dtypes)
```

```python
for company, df in stock_data.items():

    # Convert Date into datetime format
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

    # Remove rows where important stock values are missing
    df.dropna(subset=["Date", "Open", "High", "Low",
                      "Close", "Adj Close", "Volume"], inplace=True)

    # Sort data by date
    df.sort_values("Date", inplace=True)

    # Reset row index
    df.reset_index(drop=True, inplace=True)

    print(company, "cleaned successfully!",
          "| Remaining Rows:", len(df))

print("\nAll 10 companies cleaned successfully!")
```

```python
# Calculate Daily Returns and Moving Averages

for company, df in stock_data.items():

    # Daily percentage returns
    df["Daily_Return"] = df["Adj Close"].pct_change() * 100

    # 20-day Moving Average
    df["Moving_Average_20_Days"] = df["Close"].rolling(window=20).mean()

    # 50-day Moving Average
    df["Moving_Average_50_Days"] = df["Close"].rolling(window=50).mean()

    print(company, "analysis columns added!")

print("\nDaily Returns and Moving Averages calculated!")
```

```python
# Calculate 20-day rolling volatility

for company, df in stock_data.items():

    df["Volatility"] = df["Daily_Return"].rolling(window=20).std()

    print(company, "volatility calculated!")

print("\nVolatility calculation completed for all companies!")
```

```python
# Create Stock Market Performance Summary

summary_data = []

for company, df in stock_data.items():

    summary_data.append({
        "Company": company,
        "Average Closing Price": df["Close"].mean(),
        "Highest Price": df["High"].max(),
        "Lowest Price": df["Low"].min(),
        "Average Daily Return (%)": df["Daily_Return"].mean(),
        "Average Volatility (%)": df["Volatility"].mean(),
        "Total Trading Volume": df["Volume"].sum()
    })

# Convert summary into DataFrame
stock_summary = pd.DataFrame(summary_data)

# Round decimal values
stock_summary = stock_summary.round(2)

# Display summary
display(stock_summary)
```

```python
# Calculate 50-Day Volatility for All Companies

for company, df in stock_data.items():

    df["Volatility_50_Days"] = (
        df["Daily_Return"].rolling(window=50).std()
    )

    print(company, "50-Day Volatility calculated!")

print("\n50-Day Volatility completed for all companies!")
```

```python

# 1. Trading Volume Comparison - Professional Bar Chart

import matplotlib.pyplot as plt
import pandas as pd

total_volume = {
    company: df["Volume"].sum()
    for company, df in stock_data.items()
}

volume_df = pd.DataFrame(
    list(total_volume.items()),
    columns=["Company", "Total Volume"]
)

volume_df = volume_df.sort_values(
    by="Total Volume", ascending=True
)

plt.figure(figsize=(12, 7))

bars = plt.barh(
    volume_df["Company"],
    volume_df["Total Volume"]
)

plt.title("Trading Volume Comparison - Top 10 Indian Companies",
          fontsize=16, fontweight="bold", pad=18)

plt.xlabel("Total Trading Volume", fontsize=12)
plt.ylabel("Companies", fontsize=12)

plt.bar_label(bars, fmt="%.0f", padding=5, fontsize=9)

plt.grid(axis="x", linestyle="--", alpha=0.3)
plt.tight_layout()

plt.savefig(
    "charts/trading_volume_bar_chart.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# 4. Returns vs Volatility - Scatter Plot

import matplotlib.pyplot as plt
import pandas as pd

scatter_data = []

for company, df in stock_data.items():
    scatter_data.append({
        "Company": company,
        "Average Returns": df["Daily_Return"].mean(),
        "Average Volatility": df["Volatility"].mean()
    })

scatter_df = pd.DataFrame(scatter_data)

plt.figure(figsize=(12, 7))

plt.scatter(
    scatter_df["Average Volatility"],
    scatter_df["Average Returns"],
    s=130,
    alpha=0.75
)

for _, row in scatter_df.iterrows():
    plt.annotate(
        row["Company"],
        (row["Average Volatility"], row["Average Returns"]),
        xytext=(6, 6),
        textcoords="offset points",
        fontsize=8
    )

plt.title(
    "Returns vs Volatility - Top 10 Indian Companies",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Average Volatility (%)")
plt.ylabel("Average Daily Returns (%)")

plt.grid(alpha=0.3, linestyle="--")
plt.tight_layout()

plt.savefig(
    "charts/returns_vs_volatility_scatter.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
